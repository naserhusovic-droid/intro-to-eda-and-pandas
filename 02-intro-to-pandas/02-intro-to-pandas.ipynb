## What is `pandas`?
---

- Not a black-and-white mammal, it stands for **Pan**el **da**ta **s**ystem
- The most universally used data analysis library for Python (for data small enough to fit in-memory)
- Created by Wes McKinney and Open Sourced by AQR Capital Management, LLC 2009
- Implemented in highly optimized Python/Cython
    - Translation: Even though Python is "slow", `pandas` gives us C-like speeds!
- Basically put Python on the map as a data science language!

## Why use `pandas`?

- `pandas` contains all the functionality we need to explore a dataset including:
    - reading in data
    - joining multiple data sets together
    - calculating descriptive statistics
    - visualization
    - and much more!
- also, `pandas`, is **fast** and can handle large data sets (as we will see)

***Remember:***

* A **`Series`** is a one-dimensional array of values **with an index** (like indices of elements in a list)
* A **`DataFrame`** is a two-dimensional array of values **with both a row and column index**
* It turns out - each column of a `DataFrame` is actually a `Series`!

We can check this row index, which by default is the numbers 0 to length-1:
