This is because `pandas` operations almost always return a **copy** of the data without changing the underlying data set.

To make an operation permanent, you need to explicitly overwrite your `DataFrame` object:

Let's limit our data to only the most recent sets, say the last 10 years.

To do this we will:

To create new columns, we can perform a vectorized operation on the entire column (not loop through its values):

# Filtering
---

Now, we can use our new column to find only the newest sets.

To filter a `DataFrame` we need:

1. A filter to apply (like a `WHERE` clause in SQL)
2. To apply this filter to the `DataFrame`

With `pandas`, even boolean operations are vectorized!

```python
lego_df['Age'] < 10
```

```python
age_filter = lego_df["Age"] < 10
```

```python
lego_df[age_filter]
```

```python
lego_df[lego_df["Age"] < 10]
```

```python
lego_filter_df = lego_df[lego_df['Age'] < 10]
lego_filter_df
```

We can save that filter into a variable, then use it to index the `DataFrame`

```python
age_filter =lego_df["Age"] < 10
```

We can also combine those steps (you'll see this a lot)

```python
lego_df[lego_df["Age"] < 10]
```

We could also save this filtered version into a new `DataFrame` variable

```python
filtered = [lego_df["Age"] < 10]
```

This is a copy, so changes to this `DataFrame` **do not affect the original `DataFrame`** but if we tried to make any changes to it we'll get a warning

```python
filtered = lego_df[lego_df["Age"] < 10].copy()
```

This warning is just a reminder that:

- `filtered` is a "slice" of the original `lego` `DataFrame`
- it is a **copy**, so any changes to `filtered` **will not affect** `lego`

To make this go away, you can be explicit about the fact that we are copying data:

```python
filtered = [lego_df["Age"] < 10].copy()
```

No warning!

### Multiple Filters

What about only looking at new sets that have a Star Wars theme?

To combine multiple filters:

1. Each filter needs to be parenthesized
2. Instead of Python's `and` we use `&` and instead of `or` we use `|`

```python
lego_df[(lego_df["Age"] < 10) & (lego_df["Theme"] == "Star Wars")]
```

```python
lego_df[(lego_df["Age"] < 10) | (lego_df["Theme"] == "Star Wars")]
```

# Sorting
---

Let's find the most expensive LEGO sets. To do this, we need to sort our data.

Sorting is done by calling `.sort_values()` on the `DataFrame` and specifying the column.

```python
lego_df.sort_values("Current_Price")
```

The default is ascending order, so to sort in descending order we specify `ascending=False`

```python
lego_df.sort_values(("Current_Price"), ascending=False)
```
